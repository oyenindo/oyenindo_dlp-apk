# Build APK Sosmed Downloader di Google Colab
Jalankan sel dari atas ke bawah. Proses build pertama bisa 20-40 menit.

```
%%bash
apt-get update -qq
apt-get install -y -qq git zip unzip openjdk-17-jdk autoconf automake libtool pkg-config \
  zlib1g-dev libncurses5-dev libncursesw5-dev libtinfo5 cmake libffi-dev libssl-dev build-essential ccache
pip install -q --upgrade pip
pip install -q buildozer cython==0.29.36 virtualenv
```

## Unggah `main.py` dan `buildozer.spec`

```
import os, shutil
from google.colab import files
os.makedirs('/content/app', exist_ok=True)
os.chdir('/content/app')
up = files.upload()   # pilih main.py dan buildozer.spec
for n in up:
    shutil.move(n, '/content/app/' + n)
print(os.listdir('/content/app'))
```

## Build (jangan tutup tab sampai selesai)

```
import os
os.environ['BUILDOZER_WARN_ON_ROOT'] = '0'
os.environ['JAVA_HOME'] = '/usr/lib/jvm/java-17-openjdk-amd64'
os.chdir('/content/app')
!yes | buildozer -v android debug
```

## Unduh APK

```
import glob
from google.colab import files
apk = glob.glob('/content/app/bin/*.apk')
print(apk)
files.download(apk[0])
```
